```python
model = YOLO(r"C:\Users\emre_\Documents\Github\YoloV12\runs\detect\train\weights\best.pt")
test_data = sv.DetectionDataset.from_yolo(
    images_directory_path=f"{dataset}/test/images",
    annotations_directory_path=f"{dataset}/test/labels",
    data_yaml_path=f"{dataset}/data.yaml"
)
```

```python
import random
%matplotlib inline
for _ in range(5):
    i = random.randint(0, len(test_data)-1)
    
    image_path, image, target = test_data[i]
    
    results = model(image, verbose=False)[0]
    detections = sv.Detections.from_ultralytics(results).with_nms()
    
    box_annotator = sv.BoxAnnotator()
    label_annotator = sv.LabelAnnotator()
    
    labels = [
        f"{results.names[detections.class_id[i]]} {detections.confidence[i]:.2f}" 
        for i in range(len(detections))
    ]
    
    annotated_image = image.copy()
    annotated_image = box_annotator.annotate(scene=annotated_image, detections=detections)
    annotated_image = label_annotator.annotate(scene=annotated_image, detections=detections, labels=labels)

    print(f"Gösterilen görüntü: {image_path}")
    sv.plot_image(annotated_image, size=(10, 10))
    
    print("\n" + "="*80 + "\n")
```
